````python
class Pet:
    mypets = []

    def __init__(self, name):
        self.name = name
        Pet.mypets.append(self)

    def get_type(self):
        return "питомец"

    def __str__(self):
        return f"У меня есть {self.get_type()}. Его зовут {self.name}"


class Dog(Pet):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def get_type(self):
        return self.breed

    def __str__(self):
        return f"У меня есть {self.breed}. Его зовут {self.name}"


class Cat(Pet):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color.strip()

    def get_type(self):
        if "пушистый" in self.color.lower():
            return f"{self.color} кот"
        return f"{self.color}, пушистый кот"

    def __str__(self):
        return f"У меня есть {self.get_type()}. Его зовут {self.name}"


pets = [
    Dog("Бобик", "Джек Рассел"),
    Dog("Барбос", "дворняга"),
    Cat("Мурчик", "серый"),
    Pet("Тест"),
]

for pet in pets:
    print(pet)
````

## Объяснение

### Базовый класс `Pet`

Класс `Pet` является родительским для всех домашних питомцев. Атрибут класса `mypets = []` — общий список, в который автоматически добавляется каждый созданный экземпляр через `Pet.mypets.append(self)` в инициализаторе. Метод `get_type()` возвращает строку `"питомец"` — общее описание типа. Метод `__str__` формирует строку через вызов `self.get_type()`, что обеспечивает полиморфное поведение: каждый дочерний класс может переопределить `get_type()` и автоматически изменить вывод.

### Дочерний класс `Dog`

Класс `Dog` наследуется от `Pet`. Инициализатор принимает `name` и `breed` (порода). Через `super().__init__(name)` вызывается инициализатор `Pet`, который сохраняет кличку и добавляет объект в `mypets`. Затем сохраняется собственный атрибут `breed`. Метод `get_type()` переопределён — возвращает породу собаки. Метод `__str__` формирует строку с породой и кличкой.

### Дочерний класс `Cat`

Класс `Cat` наследуется от `Pet`. Инициализатор принимает `name` и `color` (цвет), причём `color` очищается от лишних пробелов через `strip()`. Метод `get_type()` проверяет, содержит ли `color` слово `"пушистый"`: если да — возвращает `"{color} кот"`, если нет — добавляет `", пушистый"` автоматически. Таким образом, достаточно передать просто `"серый"`, а `get_type()` сам сформирует `"серый, пушистый кот"`.

### Создание объектов и вывод в цикле

Создаётся список `pets` из четырёх объектов: две собаки, один кот и один базовый питомец `Pet("Тест")`. В цикле `for` вызывается `print(pet)`, который неявно вызывает `__str__`. Благодаря полиморфизму через `get_type()`, каждый объект выводит своё корректное описание.
