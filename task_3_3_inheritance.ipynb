````python
class Pet:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"У меня есть питомец. Его зовут {self.name}"


class Dog(Pet):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def __str__(self):
        return f"У меня есть {self.breed}. Его зовут {self.name}"


class Cat(Pet):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color

    def __str__(self):
        return f"У меня есть {self.color} кот. Его зовут {self.name}"


pets = [
    Dog("Бобик", "Джек Рассел"),
    Dog("Барбос", "дворняга"),
    Cat("Мурчик", "серый, пушистый"),
]

for pet in pets:
    print(pet)
````

## Объяснение

### Базовый класс `Pet`

Класс `Pet` является родительским для всех домашних питомцев. Он принимает единственный обязательный параметр `name` (кличка) и сохраняет его как атрибут экземпляра. Метод `__str__` возвращает общую строку с упоминанием питомца и его имени.

### Дочерний класс `Dog`

Класс `Dog` наследуется от `Pet`. Его инициализатор принимает два параметра: `name` и `breed` (порода). Через `super().__init__(name)` вызывается инициализатор родительского класса `Pet`, который устанавливает атрибут `name`. Затем устанавливается собственный атрибут `breed`. Метод `__str__` переопределён так, чтобы при печати объекта выводилась порода и кличка в формате: *«У меня есть {порода}. Его зовут {кличка}»*.

### Дочерний класс `Cat`

Класс `Cat` также наследуется от `Pet`. Его инициализатор принимает `name` и `color` (описание внешности/цвета). Аналогично классу `Dog`, через `super().__init__(name)` инициализируется родительский класс, а затем сохраняется атрибут `color`. Метод `__str__` переопределён для вывода в формате: *«У меня есть {цвет} кот. Его зовут {кличка}»*.

### Создание объектов и вывод в цикле

Создаётся список `pets`, содержащий три объекта: две собаки и одного кота. Далее в цикле `for` для каждого элемента списка вызывается `print(pet)`. Функция `print` неявно вызывает метод `__str__` каждого объекта. Благодаря полиморфизму, несмотря на то что все элементы списка имеют общего родителя `Pet`, каждый объект использует свою реализацию `__str__` — ту, что определена в его собственном классе (`Dog` или `Cat`).
